```python
def process_mat_file(file_path, label, signal_name="AN3", window_size=10000):
    
    # Charger le fichier .mat
    data = loadmat(file_path)
    
    # Récupérer le signal AN3
    signal = data[signal_name].flatten()
    
    # Découper le signal en fenêtres
    windows = [
        signal[i:i + window_size]
        for i in range(0, len(signal), window_size)
        if len(signal[i:i + window_size]) == window_size
    ]
    
    # Extraire les features de chaque fenêtre
    features_list = []

    for window in windows:
        features = extract_features(window)
        features_list.append(features)
    
    # Transformer en DataFrame
    df = pd.DataFrame(features_list)
    
    # Ajouter le label
    df["label"] = label
    
    return df
```

```python
import os

all_data = []

# ==========================
# HEALTHY
# ==========================

healthy_path = "../data/raw/healthy"

for file in sorted(os.listdir(healthy_path)):
    
    if file.endswith(".mat"):
        
        file_path = os.path.join(healthy_path, file)
        
        print("Processing Healthy:", file)
        
        df = process_mat_file(
            file_path,
            label=0
        )
        
        df["file"] = file
        
        all_data.append(df)


# ==========================
# DAMAGED
# ==========================

damaged_path = "../data/raw/damaged"

for file in sorted(os.listdir(damaged_path)):
    
    if file.endswith(".mat"):
        
        file_path = os.path.join(damaged_path, file)
        
        print("Processing Damaged:", file)
        
        df = process_mat_file(
            file_path,
            label=1
        )
        
        df["file"] = file
        
        all_data.append(df)
```
